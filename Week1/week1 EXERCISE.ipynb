```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python

# Get GPT-4o-mini to answer, with streaming

import os
from openai import OpenAI
from IPython.display import display, Markdown

# OpenRouter client
openrouter = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ["OPENROUTER_API_KEY"]
)

def ask_gpt_stream(question):
    response = openrouter.chat.completions.create(
        model="openai/gpt-4o-mini",
        messages=[
            {
                "role": "system",
                "content": """
                You are a friendly personal tutor.
                Explain concepts clearly and step by step.
                Use simple language, practical examples,
                and code examples when appropriate.
                """
            },
            {"role": "user", "content": question}
        ],
        temperature=0.6,
        max_tokens=1500,
        stream=True
    )

    full_response = ""
    output = display(Markdown(""), display_id=True)

    for chunk in response:
        if not chunk.choices:
            continue

        content = chunk.choices[0].delta.content

        if content:
            full_response += content
            output.update(Markdown(full_response))

    return full_response
```

```python
answer = ask_gpt_stream(
    "Explain the Transformer architecture using a simple example."
)
```

```python
# Get Llama 3.2 to answer

# Get Llama 3.2 to answer, with streaming

import ollama
from IPython.display import display, Markdown

# Ollama runs Llama locally
def ask_llama_stream(question):
    response = ollama.chat(
        model="llama3.2:1B",
        messages=[
            {
                "role": "system",
                "content": """
                You are a friendly personal tutor.
                Explain concepts clearly and step by step.
                Use simple language, practical examples,
                and code examples when appropriate.
                """
            },
            {"role": "user", "content": question}
        ],
        stream=True
    )

    full_response = ""
    output = display(Markdown(""), display_id=True)

    for chunk in response:
        content = chunk["message"]["content"]

        if content:
            full_response += content
            output.update(Markdown(full_response))

    return full_response
```

```python
answer = ask_llama_stream(
    "Explain the Transformer architecture using a simple example."
)
```

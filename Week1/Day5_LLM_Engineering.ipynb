```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

def scrape_links(url):
    response = requests.get(
        url,
        headers={"User-Agent": "Mozilla/5.0"},
        timeout=15
    )
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    links = {
        urljoin(url, a["href"])
        for a in soup.find_all("a", href=True)
    }

    return list(links)
```

```python
url = "https://www.indiatoday.in/"
print(scrape_links(url))
```

```python
from playwright.async_api import async_playwright

async def extract_links_async(url):
    async with async_playwright() as p:
        # Launch a headless browser
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()

        # Navigate to your portfolio and wait for the network to settle
        await page.goto(url, wait_until="networkidle")

        # Find all anchor tags
        anchors = await page.locator("a").all()

        # Extract 'href' attributes
        links = []
        for anchor in anchors:
            href = await anchor.get_attribute("href")
            if href:
                links.append(href)

        await browser.close()
        return links

# In a Jupyter Notebook/Colab cell, you can await the function directly:
all_links = await extract_links_async("https://chaitanya-portfolio-ten.vercel.app/")
print(all_links)
```

```python
model="nvidia/nemotron-3-super-120b-a12b:free"
```

```python
link_system_prompt = """
You are provided with a list of links found on a webpage.
You are able to decide which of the links would be most relevant to include in a brochure about the company,
such as links to an About page, or a Company page, or Careers/Jobs pages.
You should respond in JSON as in this example:

{
    "links": [
        {"type": "about page", "url": "https://full.url/goes/here/about"},
        {"type": "careers page", "url": "https://another.full.url/careers"}
    ]
}
"""
```

```python
def get_links_user_prompt(url):
    user_prompt = f"""
Here is the list of links on the website {url} -
Please decide which of these are relevant web links for a brochure about the company,
respond with the full https URL in JSON format.
Do not include Terms of Service, Privacy, email links.

Links (some might be relative links):

"""
    links = extract_links_async(url)
    user_prompt += "\n".join(links)
    return user_prompt
```

```python

import json
from openai import OpenAI
from google.colab import userdata

# Initialize OpenRouter client
openai = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=userdata.get("OPENROUTER_API_KEY")
)

# Prepare the user prompt using Playwright-extracted links
async def get_links_user_prompt(url):
    links = await extract_links_async(url)

    if links is None:
        raise ValueError(
            "extract_links_async() returned None. "
            "Check your Playwright extraction function."
        )

    if not isinstance(links, (list, tuple)):
        raise TypeError(
            f"Expected a list of links, got {type(links).__name__}"
        )

    return (
        f"Website URL: {url}\n\n"
        "Select the links most relevant to creating a detailed "
        "summary of this website. Return valid JSON with a "
        "'links' array containing the selected URLs.\n\n"
        "Available links:\n"
        + "\n".join(str(link) for link in links)
    )


# Select relevant links using OpenRouter
async def select_relevant_links(url):
    user_prompt = await get_links_user_prompt(url)

    response = openai.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": link_system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        response_format={"type": "json_object"}
    )

    result = response.choices[0].message.content

    if not result:
        raise ValueError("The model returned an empty response.")

    data = json.loads(result)

    if not isinstance(data.get("links"), list):
        raise ValueError(
            "Model output must contain a 'links' list."
        )

    print(f"Found {len(data['links'])} relevant links")
    return data


# Test on your portfolio in Google Colab
selected_links = await select_relevant_links(
    "https://chaitanya-portfolio-ten.vercel.app/"
)

print(json.dumps(selected_links, indent=2))
```

```python
from bs4 import BeautifulSoup
import requests


# Standard headers to fetch a website
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36"
}


def fetch_website_contents(url):
    """
    Return the title and contents of the website at the given url;
    truncate to 2,000 characters as a sensible limit
    """
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.content, "html.parser")
    title = soup.title.string if soup.title else "No title found"
    if soup.body:
        for irrelevant in soup.body(["script", "style", "img", "input"]):
            irrelevant.decompose()
        text = soup.body.get_text(separator="\n", strip=True)
    else:
        text = ""
    return (title + "\n\n" + text)[:2_000]


def fetch_website_links(url):
    """
    Return the links on the webiste at the given url
    I realize this is inefficient as we're parsing twice! This is to keep the code in the lab simple.
    Feel free to use a class and optimize it!
    """
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.content, "html.parser")
    links = [link.get("href") for link in soup.find_all("a")]
    return [link for link in links if link]
```

```python

async def fetch_page_and_all_relevant_links(url):
    # Fetch landing page (synchronous function)
    contents = fetch_website_contents(url)

    # Select relevant links (asynchronous function)
    relevant_links = await select_relevant_links(url)

    result = f"## Landing Page:\n\n{contents}\n\n## Relevant Links:\n"

    for link in relevant_links["links"]:
        # Support both dictionary and string link formats
        if isinstance(link, dict):
            link_url = link["url"]
            link_type = link.get("type", "Relevant Page")
        else:
            link_url = link
            link_type = "Relevant Page"

        result += f"\n\n### Link: {link_type}\n"
        result += f"URL: {link_url}\n\n"

        # Fetch linked page (synchronous function)
        result += fetch_website_contents(link_url)

    return result


# Run in Google Colab
result = await fetch_page_and_all_relevant_links(
    "https://huggingface.co"
)

print(result)
```

```python
brochure_system_prompt = """
You are an assistant that analyzes the contents of several relevant pages from a company website
and creates a short brochure about the company for prospective customers, investors and recruits.
Respond in markdown without code blocks.
Include details of company culture, customers and careers/jobs if you have the information.
"""
```

```python
async def get_brochure_user_prompt(company_name, url):
    user_prompt = f"""
You are looking at a company called: {company_name}

Here are the contents of its landing page and other relevant pages.
Use this information to build a short, engaging company brochure
in Markdown, without code blocks.

"""
    # Fetch website content asynchronously
    contents = await fetch_page_and_all_relevant_links(url)

    user_prompt += contents

    # Truncate if more than 5,000 characters
    user_prompt = user_prompt[:5000]

    return user_prompt
```

```python
prompt = await get_brochure_user_prompt(
    "HuggingFace",
    "https://huggingface.co"
)

print(prompt)
```

```python

from IPython.display import Markdown, display
import asyncio

async def create_brochure(company_name, url):
    # Generate the brochure prompt asynchronously
    prompt = await get_brochure_user_prompt(company_name, url)

    # Generate the brochure using OpenRouter
    response = await asyncio.to_thread(
        openai.chat.completions.create,
        model=model,
        messages=[
            {"role": "system", "content": brochure_system_prompt},
            {"role": "user", "content": prompt}
        ],
    )

    result = response.choices[0].message.content

    # Display the brochure as Markdown
    display(Markdown(result))

    return result


# Generate the company brochure
await create_brochure(
    "HuggingFace",
    "https://huggingface.co"
)
```

```python

import asyncio
import nest_asyncio

nest_asyncio.apply()

def run_async(coro):
    return asyncio.run(coro)
```

```python

from IPython.display import Markdown, display, update_display

def stream_brochure(company_name, url):
    prompt = run_async(
        get_brochure_user_prompt(company_name, url)
    )

    stream = openai.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[
            {"role": "system", "content": brochure_system_prompt},
            {"role": "user", "content": prompt}
        ],
        max_tokens = 5000,
        stream=True
    )

    response = ""
    display_handle = display(Markdown(""), display_id=True)

    for chunk in stream:
        if chunk.choices:
            response += chunk.choices[0].delta.content or ""
            update_display(
                Markdown(response),
                display_id=display_handle.display_id
            )

    return response


# Normal synchronous call
stream_brochure("HuggingFace", "https://huggingface.co")
```

```python

from google.colab import userdata
from openai import OpenAI

def get_fun_fact():
    api_key = userdata.get("OPENROUTER_API_KEY")

    if not api_key:
        raise ValueError(
            "OPENROUTER_API_KEY is missing. "
            "Please configure it in your environment."
        )

    openrouter = OpenAI(
        base_url="https://openrouter.ai/api/v1",
        api_key=api_key,
    )

    response = openrouter.chat.completions.create(
        model="nvidia/nemotron-3.5-lightning:free",
        messages=[
            {
                "role": "system",
                "content": (
                    "You are a friendly and engaging AI assistant. "
                    "Share accurate, interesting fun facts in simple language."
                ),
            },
            {
                "role": "user",
                "content": "Tell me one surprising fun fact and explain why it is interesting.",
            },
        ],
    )

    result = response.choices[0].message.content

    if not result:
        raise ValueError("The AI model returned an empty response.")

    return result


try:
    print(get_fun_fact())
except Exception as e:
    print(f"Error: {type(e).__name__}: {e}")
```

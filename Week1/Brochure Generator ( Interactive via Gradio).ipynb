```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

def scrape_links(url):
    response = requests.get(
        url,
        headers={"User-Agent": "Mozilla/5.0"},
        timeout=15
    )
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    links = {
        urljoin(url, a["href"])
        for a in soup.find_all("a", href=True)
    }

    return list(links)
```

```python
url = "https://www.indiatoday.in/"
print(scrape_links(url))
```

```python
from playwright.async_api import async_playwright

async def extract_links_async(url):
    async with async_playwright() as p:
        # Launch a headless browser
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()

        # Navigate to your portfolio and wait for the network to settle
        await page.goto(url, wait_until="networkidle")

        # Find all anchor tags
        anchors = await page.locator("a").all()

        # Extract 'href' attributes
        links = []
        for anchor in anchors:
            href = await anchor.get_attribute("href")
            if href:
                links.append(href)

        await browser.close()
        return links

# In a Jupyter Notebook/Colab cell, you can await the function directly:
all_links = await extract_links_async("https://chaitanya-portfolio-ten.vercel.app/")
print(all_links)
```

```python
link_system_prompt = """
You are provided with a list of links found on a webpage.
You are able to decide which of the links would be most relevant to include in a brochure about the company,
such as links to an About page, or a Company page, or Careers/Jobs pages.
You should respond in JSON as in this example:

{
    "links": [
        {"type": "about page", "url": "https://full.url/goes/here/about"},
        {"type": "careers page", "url": "https://another.full.url/careers"}
    ]
}
"""
```

```python
def get_links_user_prompt(url):
    user_prompt = f"""
Here is the list of links on the website {url} -
Please decide which of these are relevant web links for a brochure about the company,
respond with the full https URL in JSON format.
Do not include Terms of Service, Privacy, email links.

Links (some might be relative links):

"""
    links = extract_links_async(url)
    user_prompt += "\n".join(links)
    return user_prompt
```

```python

import json
from openai import OpenAI
from google.colab import userdata

# Initialize OpenRouter client
openai = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=userdata.get("OPENROUTER_API_KEY")
)

# Prepare the user prompt using Playwright-extracted links
async def get_links_user_prompt(url):
    links = await extract_links_async(url)

    if links is None:
        raise ValueError(
            "extract_links_async() returned None. "
            "Check your Playwright extraction function."
        )

    if not isinstance(links, (list, tuple)):
        raise TypeError(
            f"Expected a list of links, got {type(links).__name__}"
        )

    return (
        f"Website URL: {url}\n\n"
        "Select the links most relevant to creating a detailed "
        "summary of this website. Return valid JSON with a "
        "'links' array containing the selected URLs.\n\n"
        "Available links:\n"
        + "\n".join(str(link) for link in links)
    )


# Select relevant links using OpenRouter
async def select_relevant_links(url):
    user_prompt = await get_links_user_prompt(url)

    response = openai.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": link_system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        response_format={"type": "json_object"}
    )

    result = response.choices[0].message.content

    if not result:
        raise ValueError("The model returned an empty response.")

    data = json.loads(result)

    if not isinstance(data.get("links"), list):
        raise ValueError(
            "Model output must contain a 'links' list."
        )

    print(f"Found {len(data['links'])} relevant links")
    return data


# Test on your portfolio in Google Colab
selected_links = await select_relevant_links(
    "https://chaitanya-portfolio-ten.vercel.app/"
)

print(json.dumps(selected_links, indent=2))
```

```python
from bs4 import BeautifulSoup
import requests


# Standard headers to fetch a website
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36"
}


def fetch_website_contents(url):
    """
    Return the title and contents of the website at the given url;
    truncate to 2,000 characters as a sensible limit
    """
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.content, "html.parser")
    title = soup.title.string if soup.title else "No title found"
    if soup.body:
        for irrelevant in soup.body(["script", "style", "img", "input"]):
            irrelevant.decompose()
        text = soup.body.get_text(separator="\n", strip=True)
    else:
        text = ""
    return (title + "\n\n" + text)[:2_000]


def fetch_website_links(url):
    """
    Return the links on the webiste at the given url
    I realize this is inefficient as we're parsing twice! This is to keep the code in the lab simple.
    Feel free to use a class and optimize it!
    """
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.content, "html.parser")
    links = [link.get("href") for link in soup.find_all("a")]
    return [link for link in links if link]
```

```python

import re
from urllib.parse import urlparse, urljoin

async def fetch_page_and_all_relevant_links(url):

    result = fetch_website_contents(url)

    relevant_links = await extract_links_async(url)

    print(f"Found {len(relevant_links)} relevant links")

    base_domain = urlparse(url).hostname

    for link in relevant_links:

        # Handle either a dictionary or a plain URL
        link_url = link.get("url", "") if isinstance(link, dict) else link

        if not link_url:
            continue

        link_url = link_url.strip()

        # Remove Markdown link formatting, if present
        match = re.match(r"\[([^\]]+)\]\((https?://[^)]+)\)", link_url)

        if match:
            link_url = match.group(2)

        # Resolve relative URLs
        link_url = urljoin(url, link_url)

        parsed = urlparse(link_url)

        # Only fetch HTTP/HTTPS webpages
        if parsed.scheme not in ("http", "https"):
            print(f"Skipping non-web link: {link_url}")
            continue

        # Only fetch pages belonging to the same website
        if parsed.hostname != base_domain:
            print(f"Skipping external link: {link_url}")
            continue

        try:
            print(f"Fetching: {link_url}")
            result += "\n\n" + fetch_website_contents(link_url)

        except Exception as e:
            print(f"Skipping {link_url}: {e}")

    return result
```

```python
brochure_system_prompt = """
You are an assistant that analyzes the contents of several relevant pages from a company website
and creates a short brochure about the company for prospective customers, investors and recruits.
Respond in markdown without code blocks.
Include details of company culture, customers and careers/jobs if you have the information.
"""
```

```python
async def get_brochure_user_prompt(company_name, url):
    user_prompt = f"""
You are looking at a company called: {company_name}

Here are the contents of its landing page and other relevant pages.
Use this information to build a short, engaging company brochure
in Markdown, without code blocks.

"""
    # Fetch website content asynchronously
    contents = await fetch_page_and_all_relevant_links(url)

    user_prompt += contents

    # Truncate if more than 5,000 characters
    user_prompt = user_prompt[:5000]

    return user_prompt
```

```python

from IPython.display import Markdown, display
import asyncio

async def create_brochure(company_name, url):
    # Generate the brochure prompt asynchronously
    prompt = await get_brochure_user_prompt(company_name, url)

    # Generate the brochure using OpenRouter
    response = await asyncio.to_thread(
        openai.chat.completions.create,
        model=model,
        messages=[
            {"role": "system", "content": brochure_system_prompt},
            {"role": "user", "content": prompt}
        ],
    )

    result = response.choices[0].message.content

    # Display the brochure as Markdown
    display(Markdown(result))

    return result


# Generate the company brochure
await create_brochure(
    "HuggingFace",
    "https://huggingface.co"
)
```

```python

import asyncio
import nest_asyncio

nest_asyncio.apply()

def run_async(coro):
    return asyncio.run(coro)
```

```python

from IPython.display import Markdown, display, update_display

def stream_brochure(company_name, url):
    prompt = run_async(
        get_brochure_user_prompt(company_name, url)
    )

    stream = openai.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": brochure_system_prompt},
            {"role": "user", "content": prompt}
        ],
        max_tokens = 5000,
        stream=True
    )

    response = ""
    display_handle = display(Markdown(""), display_id=True)

    for chunk in stream:
        if chunk.choices:
            response += chunk.choices[0].delta.content or ""
            update_display(
                Markdown(response),
                display_id=display_handle.display_id
            )

    return response


# Normal synchronous call
stream_brochure("HuggingFace", "https://huggingface.co")
```

```python
import gradio as gr
import asyncio
import nest_asyncio
import tempfile
import os
import re

from urllib.parse import urlparse
from reportlab.lib import colors
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import mm
from reportlab.platypus import (
    SimpleDocTemplate,
    Paragraph,
    Spacer,
    HRFlowable
)
from xml.sax.saxutils import escape

nest_asyncio.apply()


def normalize_url(url):
    url = url.strip()

    if not url.startswith(("https://", "http://")):
        url = "https://" + url

    parsed = urlparse(url)

    if not parsed.hostname:
        raise ValueError("Please enter a valid website URL.")

    return url


def create_brochure_pdf(markdown_text, company_name):

    pdf_path = tempfile.NamedTemporaryFile(
        suffix=".pdf",
        delete=False
    ).name

    doc = SimpleDocTemplate(
        pdf_path,
        pagesize=A4,
        rightMargin=20*mm,
        leftMargin=20*mm,
        topMargin=20*mm,
        bottomMargin=20*mm
    )

    styles = getSampleStyleSheet()

    styles.add(ParagraphStyle(
        name="BrochureTitle",
        parent=styles["Title"],
        fontSize=24,
        leading=30,
        textColor=colors.HexColor("#243B53"),
        spaceAfter=18
    ))

    styles.add(ParagraphStyle(
        name="SectionHeading",
        parent=styles["Heading2"],
        fontSize=15,
        leading=20,
        textColor=colors.HexColor("#176B87"),
        spaceBefore=14,
        spaceAfter=8
    ))

    styles.add(ParagraphStyle(
        name="BodyTextCustom",
        parent=styles["BodyText"],
        fontSize=10,
        leading=15,
        spaceAfter=8
    ))

    story = [
        Paragraph(escape(company_name), styles["BrochureTitle"]),
        HRFlowable(width="100%", color=colors.HexColor("#176B87")),
        Spacer(1, 15)
    ]

    for line in markdown_text.splitlines():

        line = line.strip()

        if not line:
            story.append(Spacer(1, 5))
            continue

        if line.startswith("#"):
            heading = line.lstrip("#").strip()
            story.append(
                Paragraph(escape(heading), styles["SectionHeading"])
            )

        elif line.startswith(("- ", "* ")):
            bullet = escape(line[2:])
            story.append(
                Paragraph(
                    "• " + bullet,
                    styles["BodyTextCustom"]
                )
            )

        else:
            safe_line = escape(line)

            # Basic Markdown bold and italic support
            safe_line = re.sub(
                r"\*\*(.*?)\*\*",
                r"<b>\1</b>",
                safe_line
            )

            safe_line = re.sub(
                r"\*(.*?)\*",
                r"<i>\1</i>",
                safe_line
            )

            story.append(
                Paragraph(safe_line, styles["BodyTextCustom"])
            )

    doc.build(story)

    return pdf_path


def generate_for_gradio(url):

    try:
        url = normalize_url(url)

        company_name = (
            urlparse(url).hostname
            .replace("www.", "")
            .split(".")[0]
            .replace("-", " ")
            .title()
        )

        # Reuse your Day 5 prompt-building function
        prompt = asyncio.run(
            get_brochure_user_prompt(company_name, url)
        )

        # Reuse your existing OpenRouter client and model
        response = openai.chat.completions.create(
            model=model,
            messages=[
                {
                    "role": "system",
                    "content": brochure_system_prompt
                },
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            max_tokens=5000
        )

        brochure = response.choices[0].message.content

        if not brochure:
            raise ValueError("Nemotron returned an empty brochure.")

        pdf_path = create_brochure_pdf(
            brochure,
            company_name
        )

        return (
            brochure,
            pdf_path,
            "Brochure generated successfully!"
        )

    except Exception as e:
        return (
            "",
            None,
            f"Error: {str(e)}"
        )
```

```python
with gr.Blocks(
    theme=gr.themes.Soft(),
    title="Website Brochure AI"
) as demo:

    gr.Markdown(
        """
        # 🌐 Website Brochure AI

        ### Transform any website into a professional company brochure.

        Powered by NVIDIA Nemotron and OpenRouter.
        """
    )

    with gr.Row():
        website_url = gr.Textbox(
            label="Website URL",
            placeholder="https://example.com",
            scale=4
        )

        generate_btn = gr.Button(
            "Generate Brochure",
            variant="primary",
            scale=1
        )

    status = gr.Markdown()

    with gr.Column():
        brochure_output = gr.Markdown(
            label="Generated Brochure"
        )

        pdf_output = gr.File(
            label="Download Your PDF Brochure"
        )

    generate_btn.click(
        fn=generate_for_gradio,
        inputs=website_url,
        outputs=[
            brochure_output,
            pdf_output,
            status
        ]
    )

    website_url.submit(
        fn=generate_for_gradio,
        inputs=website_url,
        outputs=[
            brochure_output,
            pdf_output,
            status
        ]
    )

demo.launch(share=True, debug=True)
```

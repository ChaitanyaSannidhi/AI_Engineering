```python
import os
from openai import OpenAI

# Get your API key from the environment
OPENROUTER_API_KEY = userdata.get("OPENROUTER_API_KEY")

if not OPENROUTER_API_KEY:
    raise ValueError("OPENROUTER_API_KEY is not set")

# Create the OpenRouter client
openrouter = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=OPENROUTER_API_KEY
)

# Select your model
MODEL = "nvidia/nemotron-3.5-lightning:free"

# Define your system prompt
system_message = """
You are a friendly personal AI tutor.
Explain concepts step by step using simple language
and practical examples. Include code examples when helpful.
"""

print("OpenRouter client initialized successfully!")
```

```python
import gradio as gr
from openai import OpenAI

# Your existing OpenRouter client and variables:
# openrouter = OpenAI(
#     base_url="https://openrouter.ai/api/v1",
#     api_key=OPENROUTER_API_KEY
# )
#
# MODEL = "your-model-name"
# system_message = "You are a helpful assistant."

def chat(message, history):
    try:
        # Convert Gradio history into OpenAI message format
        messages = [
            {"role": "system", "content": system_message}
        ]

        for h in history:
            if h.get("role") in ("user", "assistant"):
                messages.append({
                    "role": h["role"],
                    "content": h["content"]
                })

        # Add the latest user message
        messages.append({
            "role": "user",
            "content": message
        })

        response = openrouter.chat.completions.create(
            model=MODEL,
            messages=messages
        )

        answer = response.choices[0].message.content

        if answer is None:
            return "The model returned no text. Check its reasoning or response format."

        return answer

    except Exception as e:
        return f"Error: {type(e).__name__}: {e}"

gr.ChatInterface(
    fn=chat,
).launch()
```

```python
def chat(message, history):
    history = [{"role":h["role"], "content":h["content"]} for h in history]
    messages = [{"role": "system", "content": system_message}] + history + [{"role": "user", "content": message}]
    stream = openrouter.chat.completions.create(model=MODEL, reasoning_effort="none", messages=messages, stream=True)
    response = ""
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        yield response
```

```python
gr.ChatInterface(fn=chat).launch()
```

```python
def chat(message, history):
    history = [{"role":h["role"], "content":h["content"]} for h in history]
    relevant_system_message = system_message
    if 'belt' in message.lower():
        relevant_system_message += " The store does not sell belts; if you are asked for belts, be sure to point out other items on sale."

    messages = [{"role": "system", "content": relevant_system_message}] + history + [{"role": "user", "content": message}]

    stream = openrouter.chat.completions.create(model=MODEL, reasoning_effort="none", messages=messages, stream=True)

    response = ""
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        yield response
```

```python
gr.ChatInterface(fn=chat).launch()
```
